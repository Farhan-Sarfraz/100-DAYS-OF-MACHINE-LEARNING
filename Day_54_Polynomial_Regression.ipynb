```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
np.random.seed(42)

X = np.random.uniform(-5, 5, 1000).reshape(-1,1)

y = 3*(X**2) + 2*X + 5 + np.random.randn(1000,1)*5
```

```python
plt.figure(figsize=(8,5))

plt.scatter(X, y)

plt.title("Random Non-Linear Dataset")
plt.xlabel("X")
plt.ylabel("y")

plt.show()
```

# By Linear Reg

```python
linear_model = LinearRegression()

linear_model.fit(X, y)
```

```python
y_pred_linear = linear_model.predict(X)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(X, y, label="Actual Data")

plt.plot(X, y_pred_linear, color='red', linewidth=2,
         label="Linear Regression")

plt.title("Simple Linear Regression")
plt.xlabel("X")
plt.ylabel("y")

plt.legend()

plt.show()
```

# By Polynomial Regression

```python
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)
```

```python
poly_model = LinearRegression()

poly_model.fit(X_poly, y)
```

```python
y_pred_poly = poly_model.predict(X_poly)
```

```python
sorted_index = np.argsort(X.flatten())

X_sorted = X[sorted_index]

y_sorted = y_pred_poly[sorted_index]
```

```python
plt.figure(figsize=(8,5))

plt.scatter(X, y, label="Actual Data")

plt.plot(X_sorted,
         y_sorted,
         color='red',
         linewidth=3,
         label="Polynomial Regression")

plt.title("Polynomial Regression (Degree = 2)")
plt.xlabel("X")
plt.ylabel("y")

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(X, y, label="Actual Data")

# Linear Regression
plt.plot(X,
         y_pred_linear,
         color='green',
         linewidth=2,
         label="Linear Regression")

# Polynomial Regression
plt.plot(X_sorted,
         y_sorted,
         color='red',
         linewidth=3,
         label="Polynomial Regression")

plt.title("Linear vs Polynomial Regression")

plt.xlabel("X")
plt.ylabel("y")

plt.legend()

plt.show()
```

```python
from sklearn.metrics import r2_score, mean_squared_error

print("====-Linear Regression ====")

print("R² Score :", r2_score(y, y_pred_linear))

print("MSE :", mean_squared_error(y, y_pred_linear))

print()

print("==== Polynomial Regression ====")

print("R² Score :", r2_score(y, y_pred_poly))

print("MSE :", mean_squared_error(y, y_pred_poly))
```

# For 3D 

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from mpl_toolkits.mplot3d import Axes3D
```

```python
np.random.seed(42)

# Two input features
X1 = np.random.uniform(-3,3,100)
X2 = np.random.uniform(-3,3,100)

# Target (Non-linear relationship)
y = 3*(X1**2) + 2*(X2**2) + 4*X1 - 3*X2 + np.random.randn(100)*3

# Combine Features
X = np.column_stack((X1,X2))
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111,projection='3d')

ax.scatter(X1,X2,y)

ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("y")

plt.show()
```

# SIMPLE LINEAR REGRESSION

```python
lr = LinearRegression()

lr.fit(X,y)
```

```python
x1_range = np.linspace(X1.min(),X1.max(),30)

x2_range = np.linspace(X2.min(),X2.max(),30)

xx1,xx2 = np.meshgrid(x1_range,x2_range)
```

```python
grid = np.column_stack((xx1.ravel(),xx2.ravel()))

plane = lr.predict(grid)

plane = plane.reshape(xx1.shape)
```

```python
fig = plt.figure(figsize=(9,7))

ax = fig.add_subplot(111,projection='3d')

ax.scatter(X1,X2,y,label="Actual Data")

ax.plot_surface(xx1,
                xx2,
                plane,
                alpha=0.5)

ax.set_title("Simple Linear Regression")

ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("y")

plt.show()
```

# POLYNOMIAL REGRESSION

```python
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)
```

```python
poly_model = LinearRegression()

poly_model.fit(X_poly,y)
```

```python
grid_poly = poly.transform(grid)
```

```python
surface = poly_model.predict(grid_poly)

surface = surface.reshape(xx1.shape)
```

```python
fig = plt.figure(figsize=(9,7))

ax = fig.add_subplot(111,projection='3d')

ax.scatter(X1,X2,y,label="Actual Data")

ax.plot_surface(xx1,
                xx2,
                surface,
                alpha=0.5)

ax.set_title("Polynomial Regression")

ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("y")

plt.show()
```

```python
fig = plt.figure(figsize=(15,6))

# Linear Regression
ax1 = fig.add_subplot(121,projection='3d')

ax1.scatter(X1,X2,y)

ax1.plot_surface(xx1,
                 xx2,
                 plane,
                 alpha=0.5)

ax1.set_title("Linear Regression")

# Polynomial Regression
ax2 = fig.add_subplot(122,projection='3d')

ax2.scatter(X1,X2,y)

ax2.plot_surface(xx1,
                 xx2,
                 surface,
                 alpha=0.5)

ax2.set_title("Polynomial Regression")

plt.show()
```

```python
from sklearn.metrics import r2_score

linear_pred = lr.predict(X)

poly_pred = poly_model.predict(X_poly)

print("Linear Regression R² :", r2_score(y,linear_pred))

print("Polynomial Regression R² :", r2_score(y,poly_pred))
```
