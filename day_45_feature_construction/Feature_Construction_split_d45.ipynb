```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.sample(5)
```

# Feature Construction

```python
df.head()
```

```python
df.dropna(inplace = True)
# missing data in age is dopped
```

```python
df.head()
```

```python
X = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=50))
```

```python
# feature construction ---->>
```

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
# yaha 1 woh memeber add hota ha woh bhe family ka part hai 
```

```python
X.head(20)
```

```python
def myfunc(num):
    if num == 1:
        #alone aik bnda
        return 0
    elif num >1 and num <=4:
        # small family
        return 1
    elif num >= 4:
        return 2
        # large family
    else:
        print('Enter Valid Number, Please!')
```

```python
myfunc(3)
```

```python
myfunc(4)
```

```python
myfunc(9)
```

```python
myfunc(0)
```

```python
X['Family_type'] = X['Family_size'].apply(myfunc)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=50))
```
